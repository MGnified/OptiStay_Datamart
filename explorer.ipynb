```python
from faker import Faker
import pandas as pd
import random
from unidecode import unidecode

Faker.seed(27)
fake = Faker()

names = {name: fake.name() for name in range(10)}

names_df = pd.Series(names)

print(names_df)
```

```python
fake_ger = Faker("de_DE")

# specifying first and last name
first_name = fake_ger.first_name()
last_name = fake_ger.last_name()

# combining first and last name
full_name = f"{first_name} {last_name}"
print(full_name)

# creating an email based on the name
first_name_cleaned = unidecode(first_name.lower())
last_name_cleaned = unidecode(last_name.lower())
email = f"{first_name_cleaned}.{last_name_cleaned}@{fake.free_email_domain()}"
print(email)
```

```python
def generate_personal_data(locale: str):
    fake = Faker(locale)

    first_name = fake.first_name()
    last_name = fake.last_name()
    full_name = f"{first_name} {last_name}"

    email_first = unidecode(first_name.lower()).replace(" ", "")
    email_last = unidecode(last_name.lower()).replace(" ", "")

    email = f"{email_first}.{email_last}@{fake.free_email_domain()}"

    address = fake.address().replace("\n", ", ")

    return full_name, email, address


locales = [
    "de_DE",
    "de_AT",
    "de_CH",
    "it_IT",
    "fr_FR",
    "en_GB",
    "da_DK",
    "es_ES",
    "nl_NL",
    "sv_SE",
    "no_NO",
]

people_data = []

for person in range(21):
    locale = random.choice(locales)
    people_data.append(generate_personal_data(locale))


df = pd.DataFrame(people_data, columns=("name", "email", "address"))

df
```
